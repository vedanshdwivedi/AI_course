## Lemmatisation

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = ["eat", "eating", "eats", "eaten", "love", "loving", "write", "written", "writing", "handwriting", "finally", "finalised", "history", "congratulations", "going", "goes", "gone"]
```

```python
# noun - n
# verb - v
# adjective - a
# adverb - r

for word in words:
    print(f"{word} -----> {lemmatizer.lemmatize(word, pos='n')}")
```

```python
for word in words:
    print(f"{word} -----> {lemmatizer.lemmatize(word, pos='v')}")
```

```python
for word in words:
    print(f"{word} -----> {lemmatizer.lemmatize(word, pos='a')}")
```

```python
for word in words:
    print(f"{word} -----> {lemmatizer.lemmatize(word, pos='r')}")
```
