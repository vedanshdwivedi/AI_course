# Stemming

Stemming is the process of reducing a word to it's word's stem. For instance, stemming(eating) = eat, stemming(dancing) = dance. This is usually important in problem statements, where we let's say want to perform sentiment analysis of a movie review, and we want to treat "love" and "loving" as the same word.

```python
# Classification Problem
# Lets say we want to classify reviews as positive or negative. 
# Reviews -> These would be textual data. There can be words like eat, eating, eaten etc which is from the same stem word 'eat'
```

```python
words = ["eat", "eating", "eats", "eaten", "love", "loving", "write", "written", "writing", "handwriting", "finally", "finalised", "history", "congratulations"]
```

## Porter Stemmer

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```

## RegexpStemmer

This takes a regular expression and removes suffix/prefix matching the expression

```python
from nltk.stem import RegexpStemmer
regex = "ing$|s$|able$"
stemmer = RegexpStemmer(regex, 4)
```

```python
for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```

## SnowballStemmer

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer("english")
```

```python
for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```
