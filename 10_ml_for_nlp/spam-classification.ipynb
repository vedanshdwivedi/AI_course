# Spam Classification

We will use a spam classification dataset for this practical. This also will cover the following topics:
1. Bag of words
2. N-grams
3. TF-IDF
4. word2vec

```python
import nltk
import pandas as pd
import numpy as np
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
nltk.download("stopwords")
```

```python
messages = pd.read_csv("../datasets/spam.csv", encoding="latin-1")
messages = messages[["v1", "v2"]]
messages = messages.rename(columns={"v1": "label", "v2": "message"})
df = messages.copy()
messages.head()
```

```python
# Data Cleaning and Preprocessing

# Lowercase chars
# Remove special chars
# Remove stop words
# Apply stemming
 
corpus = []
for i in range(len(messages)):
    label, message = messages.iloc[i]
    lowercaseMessage = message.lower()
    cleanMessage = re.sub(r'[^a-zA-Z0-9]', ' ', lowercaseMessage)
    tokenizedWords = nltk.word_tokenize(cleanMessage)
    stemmedWords = [lemmatizer.lemmatize(x) for x in tokenizedWords]
    removedStopWords = [x for x in stemmedWords if x not in stopwords.words("english")]
    joinedSentence = " ".join(removedStopWords)
    corpus.append(joinedSentence)

df["processed_message"] = corpus
df.head()
```

```python
enableBinaryBagOfWords = True
vectorizer = CountVectorizer(max_features=500, binary=enableBinaryBagOfWords, ngram_range=(2,3))
X = vectorizer.fit_transform(corpus).toarray()
X.shape
```

```python
vectorizer.vocabulary_
```

```python
## TF-IDF
tfVectorizer = TfidfVectorizer(max_features=100, ngram_range=(2,2))
X = tfVectorizer.fit_transform(corpus).toarray()
X.shape
```
