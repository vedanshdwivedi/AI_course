# Parts of Speech tagging in NLP

```python
paragraph = """I need a DevOps engineer to make an existing Swiss healthcare app run entirely without Supabase Cloud. The app is built and tested; the self-host stack is missing Storage, Realtime, an edge-function runtime and TLS. You self-host the remaining Supabase components in Docker, vendor the edge functions so they start without internet, harden auth, and set up backups with a proven restore. Deep PostgreSQL row-level-security experience is essential — ~40 policies depend on one function that needs cleaning up in a live database. First, I would like to review this project in a Teams meeting and discuss its feasibility before the contract begins. If you are interested in co-founding the company Zeme-echt.ch and finishing the development of the other four applications, we offer a partnership agreement with a 25% profit share, scaling from the very first revenue once integrated into an operating system. There is strong interest from participants in the long-term care sector; now we just need someone ready to roll up their sleeves and dive in. Take a look at our website to get a sense of where we currently stand. We want you! A foundation has also expressed strong interest. Once the applications are integrated and ready for on-site testing, they will receive the product free of charge. In return, we gain a valuable reference and an operational platform to test everything before marketing it on a larger scale. Are you interested?
"""
```

```python
from nltk import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import nltk

nltk.download('averaged_perceptron_tagger_eng')
```

```python
sentences = sent_tokenize(paragraph)
```

```python
sentences[0]
```

```python
# Derive POS Tag for each word

word_list = []
for sentence in sentences:
    words = word_tokenize(sentence)
    words = [x for x in words if x not in set(stopwords.words('english'))]
    pos_tag = nltk.pos_tag(words)
    word_list.append(pos_tag)
```
