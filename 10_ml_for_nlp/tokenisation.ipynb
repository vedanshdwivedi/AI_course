# Tokenisation
#### We would require NLTK (https://www.nltk.org/) and Spacy (https://spacy.io/) libraries which are open-source

```python
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize, TreebankWordTokenizer
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')

tokenizer = TreebankWordTokenizer()
```

```python
corpus = """Hello Welcome, to Vedansh Dwivedi's NLP journey. 
Vedansh Dwivedi loves tech and he is now exploring AI from the beginning.
NLP is the first milestone in this journey. 
It has been amazing so far.
"""

print(corpus)
```

```python
## Tokenisation 
## Paragraph -> Sentences
documents=sent_tokenize(corpus)
documents
```

```python
# Tokenisation
# Paragraph -> Words
# Sentences -> Words
corpus_words = word_tokenize(corpus)
corpus_words[:5]
```

```python
# word_tokenize does not treat puncutations as tokens
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
# wordpunct_tokenize treats puncutations as different tokens
for sentence in documents:
    print(wordpunct_tokenize(sentence))
```

```python
# TreeBankWordTokenizer will not consider . as a different token
tokenizer.tokenize(corpus)[:11]
```
